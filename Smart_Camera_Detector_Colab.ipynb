# Smart Camera Object Detector
**Datorseende-demo i stil med Axis video-analys (edge AI)**

Detta program hittar och raknar personer/objekt i en bild – samma sorts analys som korr pa en Axis-kamera.
Du tranar ingen modell sjalv; du anvander en **fardigtranad** YOLOv8-modell.

Kor cellerna uppifran och ner med **Shift+Enter**.

## 1. Installera bibliotek (ca 30 sek)

```python
!pip install ultralytics -q
from ultralytics import YOLO
import cv2
from google.colab.patches import cv2_imshow
print('Klart!')
```

## 2. Ladda den fardigtranade modellen
YOLOv8n ar en liten, snabb modell – precis tanken med 'edge' (kor pa svag hardvara).

```python
model = YOLO('yolov8n.pt')
print('Modellen kan kanna igen:', len(model.names), 'olika typer av objekt')
```

## 3. Ladda upp en bild
Kor cellen, klicka 'Valj filer' och valj en bild med personer/bilar/objekt (t.ex. en gatubild).

```python
from google.colab import files
uploaded = files.upload()
image_path = list(uploaded.keys())[0]
print('Uppladdad bild:', image_path)
```

## 4. Kor detektionen och visa resultatet

```python
results = model(image_path)
result = results[0]

# Rita rutor och visa bilden
annotated = result.plot()
cv2_imshow(annotated)
cv2.imwrite('output.jpg', annotated)
```

## 5. Rakna objekt + enkelt sakerhetslarm

```python
names = model.names
classes = [int(c) for c in result.boxes.cls.tolist()]
counts = {}
for c in classes:
    counts[names[c]] = counts.get(names[c], 0) + 1

print('Hittade', len(classes), 'objekt:')
for label, n in sorted(counts.items(), key=lambda x: -x[1]):
    print(f'  - {label}: {n}')

people = counts.get('person', 0)
if people > 0:
    print(f'\n  INTRANGSLARM: {people} person(er) upptackt(a)!')
else:
    print('\nOK: inga personer upptackta.')
```
